# Data Cleaning

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
#Extracting.
```

```python
df['year'] = df['date_added'].dt.year
```

```python
df['month'] = df['date_added'].dt.month
```

```python
df['day'] = df['date_added'].dt.day
```

```python
#filling missing values in country.
df['country'].fillna('unknown', inplace=True)
```

```python
df['director'].fillna('unknown', inplace=True)
```

```python
df.dropna(subset=['cast','date_added','rating','duration'],inplace=True)
```

# Bivariate Analysis

```python
plt.figure(figsize=(15,7))
sns.countplot(x= df['type'], hue= df['rating'])
plt.title('Rating vs Type', size=20)
plt.show()
```

```python
movies = df[df['type'] == 'Movie'].copy()
```

```python
movies['duration'] = ( movies['duration'] .str.extract(r'(\d+)')[0].astype(float))
```

```python
avg_duration = (movies.groupby('release_year')['duration'].mean().reset_index())
```

```python
sns.lineplot(data=avg_duration, x='release_year', y='duration', marker='o')

plt.title('Average Movie Duration by Release Year', fontsize=15, fontweight='bold')
plt.xlabel('Release Year')
plt.ylabel('Average Duration (Minutes)')
plt.show()
```

# Business Insights

```python

country_df = df.assign(country=df['country'].fillna('Unknown').str.split(', '))
country_df = country_df.explode('country')

top_countries = country_df['country'].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot( x=top_countries.values, y=top_countries.index, palette='viridis')
plt.title("Top 10 Countries Producing Netflix Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")
plt.show()
```

```python
genre_df = df.assign(listed_in=df['listed_in'].str.split(', '))
genre_df = genre_df.explode('listed_in')

top_genres = genre_df['listed_in'].value_counts().head(10)

plt.figure(figsize=(12,6))
sns.barplot( x=top_genres.values, y=top_genres.index, palette='magma')
plt.title("Top 10 Genres on Netflix")
plt.show()
```

```python
movies = df[df['type']=='Movie']

movie_year = movies['release_year'].value_counts().sort_index()

plt.figure(figsize=(14,6))
sns.lineplot(x=movie_year.index, y=movie_year.values, marker='o')
plt.title("Movie Production Over Time")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.show()
```

```python
year_growth = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(14,6))
sns.lineplot(x=year_growth.index, y=year_growth.values,marker='o')
plt.title("Netflix Content Growth by Release Year")
plt.show()
```
